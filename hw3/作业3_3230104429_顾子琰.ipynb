#### 4.2 找出数据中存在的所有完全非平凡函数依赖（5分）

由于无法了解客户的需求，只能通过试错法不断尝试，从最简单的函数依赖开始，如name $\rightarrow$ discount，即当记录的name属性相同时，对应的discount属性也相同，然后再尝试复杂的函数依赖，如name, discount $\rightarrow$ month。对于每个函数依赖，构造一个SQL语句进行检查，要求SQL语句尽可能简短，且能通过查询结果是否为空集，确定是否存在所查找的函数依赖（不是看数据内容，人工判断确定）。

Be Clever，结合函数依赖的规则减少需要检查的函数依赖数目，同时不能丢失任何可能的函数依赖。

构造一个SQL查询语句，检查是否存在A $\rightarrow$ B类型的完全非平凡函数依赖，其中A和B可以是Sales关系中的任何属性（从Sales关系中选择两个不同属性，使用下面SQL语句检查是否存在这一函数依赖）。

```python
%%sql
-- 检查name->discount
select name, count(distinct discount)
from Sales
group by name
having count(distinct discount) > 1;
```

```python
%%sql
-- 检查month->discount
select month, count(distinct discount)
from Sales
group by month
having count(distinct discount) > 1;
```

```python
%%sql
-- 检查name->price
select name, count(distinct price)
from Sales
group by name
having count(distinct price) > 1;
```

构造一个SQL查询语句，检查是否存在A, B $\rightarrow$ C类型的完全非平凡函数依赖，其中A、B和C可以是Sales关系中的任何属性（从Sales关系中选三个不同属性，使用下面SQL语句检查是否存在这一函数依赖）。

```python
%%sql
-- 检查name,month->discount
select name, month, count(distinct discount)
from Sales
group by name, month
having count(distinct discount) > 1;
```

```python
%%sql
-- 检查name,discount->price
select name, discount, count(distinct price)
from Sales
group by name, discount
having count(distinct price) > 1;
```

通过上述查找，给出Sales关系中存在的最小**完全非平凡**函数依赖集合，即a minimal set of completely nontrivial FDs such that all FDs that hold on the relation follow from the dependencies in the set，不在该集合中的函数依赖，都可以通过集合中的函数依赖根据规则推导得到。单属性之间的判断可以逐个写SQL语句判断，也可以修改以下Python代码自动判断所有单属性依赖。

```python
colums = ['name', 'discount', 'month', 'price']
template = 'select %s, count(distinct %s) from Sales group by %s having count(distinct %s) > 1' # 使用%s作为占位符，类似于C语言的%s

for A in colums:
    for B in colums:
        query = template % (A, B, A, B)
        result = %sql $query
        if A != B and len(result) == 0:
            print(A + '->' + B)
```

对于Sales关系，获得上述最小完全非平凡函数依赖集合至少需要多少次SQL查询？三个属性之间的函数依赖(A, B $\rightarrow$ C)可以利用两个属性之间的函数依赖(A $\rightarrow$ B)集合减少SQL查询次数，提示不大于20。

```python
colums = ['name', 'discount', 'month', 'price']
template = 'select %s, count(distinct %s) from Sales group by %s having count(distinct %s) > 1' # 使用%s作为占位符，类似于C语言的%s
count_sql = 0

for A in colums:
    for B in colums:
        query = template % (A, B, A, B)
        count_sql += 1
        result = %sql $query
        if A != B and len(result) == 0:
            print(A + '->' + B)

print('总共执行了' + str(count_sql) + '次SQL查询')
```
